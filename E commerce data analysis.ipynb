```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Load Data (Desktop Path)
file_path = r"C:\Users\Lenovo\OneDrive\Desktop\Amazon Sale Report.csv"
df = pd.read_csv(file_path, low_memory=False)

# 2. Data Cleaning
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Amount"] = df["Amount"].fillna(0)
df["Qty"] = df["Qty"].fillna(0)
df["Category"] = df["Category"].fillna("Unknown")
df["Fulfilment"] = df["Fulfilment"].fillna("Unknown")
df["ship-state"] = df["ship-state"].fillna("Unknown")

# 3. Time Features
df["Year-Month"] = df["Date"].dt.to_period("M").astype(str)

# 4. Machine Learning: Sales Trend Forecasting
monthly_sales = (
    df.groupby("Year-Month")["Amount"].sum().reset_index().sort_values("Year-Month")
)
monthly_sales["Month_Index"] = np.arange(len(monthly_sales))

# Model Training
X = monthly_sales[["Month_Index"]]
y = monthly_sales["Amount"]

model = LinearRegression()
model.fit(X, y)

# Predict Sales
monthly_sales["Forecasted_Amount"] = model.predict(X).round(2)

# Merge predictions back to main dataset
df = df.merge(
    monthly_sales[["Year-Month", "Forecasted_Amount"]],
    on="Year-Month",
    how="left",
)

# 5. Export Cleaned Dataset for Power BI directly to Desktop
export_path = r"C:\Users\Lenovo\OneDrive\Desktop\ecommerce_cleaned.csv"
df.to_csv(export_path, index=False)

print("SUCCESS: Data Cleaned, Forecasted & Exported!")
print("Cleaned file saved at:", export_path)
```
